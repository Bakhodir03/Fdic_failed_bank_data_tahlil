```python
import pandas as pd

# HTML fayldan barcha jadvallarni ro‘yxat sifatida o‘qib oladi
df_list = pd.read_html(r'/content/fdic_failed_bank_list.html')

# Ro‘yxatdagi birinchi (asosiy) jadvalni ajratib olish
df = df_list[0]
def Info():
    # DataFrame haqidagi umumiy ma'lumot
    print(F"dataning hajmi-->\t{df.info()}")
def Columns():
    # Ustun nomlarini chiqarish
    print(df.columns)
def Head(num):
    print(f"head--> {df.head(num)}")
def Shape():
    print("Shape-->",df.shape)
    print(f"Ustunlari {df.shape[1]}")
    print(f"Qatorlari {df.shape[0]}")
def Describe():
  print(df.describe)
```

#  1. Vaqt bo‘yicha tahlil savollari

```python
#  1. Vaqt bo‘yicha tahlil savollari
#1.1 Qaysi yilda eng ko‘p bank yopilgan?
def Closing_year():
    df['Closing Date'] = pd.to_datetime(df['Closing Date'])
    df['Year'] = df['Closing Date'].dt.year
    print(df['Year'].value_counts().sort_index())
def Top_5_years():
    top_5_years = df['Year'].value_counts().head(5)
    print("Eng ko‘p bank yopilgan 5 yil:")
    print(top_5_years)

#1.2 Har yili banklar yopilish tendensiyasi qanday o‘zgarib borgan?
def Closing_tendency():
    yearly_counts = df['Year'].value_counts().sort_index()
    plt.figure(figsize=(10,6))
    plt.plot(yearly_counts.index, yearly_counts.values, marker='o', linestyle='-', color='royalblue')
    plt.title("Har yili banklar yopilish tendensiyasi", fontsize=14)
    plt.xlabel("Yil")
    plt.ylabel("Yopilgan banklar soni")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
def Closing_max():
    peak_year = df['Year'].value_counts().idxmax()
    count = df['Year'].value_counts().max()
    print(f"Eng ko‘p bank yopilgan yil: {peak_year} da {count} ta bank")

#1.3  Oylar bo‘yicha bank yopilishining eng yuqori foizi qachon?
def Closing_month_list():
    df['Closing Date'] = pd.to_datetime(df["Closing Date"])
    df["Month"] = df['Closing Date'].dt.month_name()
    return df['Month'].value_counts().sort_index()

def Closing_month_figure():
    month_counts = df['Month'].value_counts().sort_index()
    plt.figure(figsize=(10,6))
    month_counts.plot(kind='bar', color='salmon')
    plt.title("Oy bo‘yicha banklar yopilish soni")
    plt.xlabel("Oy")
    plt.ylabel("Banklar soni")
    plt.xticks(rotation=45)
    plt.grid(axis='y')
    plt.tight_layout()
    plt.show()

def Closing_month_pie():
    month_counts = df['Month'].value_counts()
    plt.figure(figsize=(7,7))
    plt.pie(data =df , x=month_counts, colors=sns.color_palette('bright'),labels=month_counts.index,autopct="%1.2f%%")
    plt.title("oy boyicaha yopilgan banklar")
    plt.axis("equal")
    plt.show()

def CLosing_2008_2010():
    year_closing = df['Year'].value_counts().sort_index()
    # 2008-2010 yillar oralig‘idagi yopilishlar soni
    crisis_years = year_closing.loc[2008:2010]
    other_years = year_closing.drop(labels=[2008,2009,2010])

    print("2008–2010 yillardagi bank yopilishi:")
    print(crisis_years)
    print("\nBoshqa yillarning o‘rtacha yopilishi:")
    print(f"O'rtacha: {other_years.mean():.2f} ta bank/ yil")
```

# 2. Hududiy tahlil savollari (ST va City ustunlari)

```python
# 2. Hududiy tahlil savollari (ST va City ustunlari)
# 2.1 Hududiy tahlil savollari (ST va City ustunlari)

def Top_closing_shtat():
    shtate_counts = df['ST'].value_counts().sort_values(ascending=False)
    print("Eng ko'p yopilgan shtat:", shtate_counts.head(1))

# 2.2 Qaysi shaharda eng ko‘p bank yopilgan?
def Closing_city():
    city_max = df['City'].value_counts().sort_values(ascending=False)
    print("Eng ko'p yopilgan shahar:", city_max.head(1))

# 2.3 Eng ko‘p yopilgan banklar 5 ta shtatda qanday taqsimlangan?
def Top_5_shtate():
    shtate_closing = df['ST'].value_counts().sort_values(ascending=False).head(5)
    return shtate_closing

# 2.4 Bank yopilishi geografik jihatdan qayerda ko‘p uchraydi?
def Region_distribution():
    # Mintaqalar ro‘yxati
    south_states = ['TX', 'FL', 'GA', 'AL', 'MS', 'LA', 'SC', 'NC', 'TN', 'AR', 'OK', 'KY']
    west_states = ['CA', 'AZ', 'NV', 'OR', 'WA', 'CO', 'NM', 'UT', 'ID', 'MT', 'WY', 'AK', 'HI']
    east_states = ['NY', 'NJ', 'MA', 'CT', 'PA', 'MD', 'DE', 'RI', 'VA', 'DC', 'ME', 'VT', 'NH']
    midwest_states = ['IL', 'OH', 'MI', 'MN', 'WI', 'MO', 'IA', 'KS', 'IN', 'ND', 'SD', 'NE']

    def assign_region(state):
        if state in south_states:
            return 'Janub'
        elif state in west_states:
            return 'G‘arb'
        elif state in east_states:
            return 'Sharq'
        elif state in midwest_states:
            return 'Markaziy'
        else:
            return 'Boshqa'

    df['Region'] = df['ST'].apply(assign_region)
    region_counts = df['Region'].value_counts()

    print("Hududlar bo‘yicha bank yopilishi taqsimoti:")
    print(region_counts)

def ensure_region():
    if 'Region' not in df.columns:
        Region_distribution()   # ustunni yaratib beradi

def Region_distribution_pie():
    ensure_region()
    region_counts = df['Region'].value_counts().sort_index()
    plt.figure(figsize=(8,8))
    plt.pie(region_counts, labels=region_counts.index, autopct='%1.1f%%',
            startangle=140, colors=plt.cm.Set2.colors,
            explode=[0.1]*len(region_counts))
    plt.title("Bank yopilishi geografik taqsimoti")
    plt.axis('equal')
    plt.tight_layout()
    plt.show()

def Region_distribution_plot():
    ensure_region()
    region_counts = df['Region'].value_counts()
    plt.figure(figsize=(8,8))
    region_counts.plot(kind='bar', color='salmon')
    plt.title('Bank yopilishi geografik taqsimoti')
    plt.ylabel("Banklar soni")
    plt.xlabel("Region")
    plt.xticks(rotation=45)
    plt.grid(axis='y')
    plt.tight_layout()
    plt.show()
```

# 4. Statistika, sonli tahlil savollari

```python
# 4.1 CERT raqami eng katta va eng kichik bo‘lgan banklar qaysilar?

def Cert_min_max_bank_name():
    cert_max = df['CERT'].value_counts().sort_values(ascending=False).head(1)
    max_cert_value = cert_max.index[0]

    cert_min = df['CERT'].value_counts().sort_values().head(1)
    min_cert_value = cert_min.index[0]

    max_cert_bank = df[df['CERT'] == max_cert_value]['Bank Name'].unique()
    min_cert_bank = df[df['CERT'] == min_cert_value]['Bank Name'].unique()

    print(f"Eng ko‘p CERT raqam:\t{max_cert_value}\t→ Bank nomi: {max_cert_bank[0]}")
    print(f"Eng kam  CERT raqam:\t{min_cert_value}\t→ Bank nomi: {min_cert_bank[0]}")
def Cert_mean():
    return float(df['CERT'].mean())

```
